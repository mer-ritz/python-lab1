```python
#Accept roll numbers enrolled in three courses as sets. Display students in all three courses, students in at least one course, students in exactly one course, students in course one but not course two, and the number of students taking exactly two courses.

A=set(map(int, input("enter roll numbers for course 1:").split()))
B=set(map(int, input("enter roll numbers for course 2:").split()))
C=set(map(int, input("enter roll numbers for course 3:").split()))

print("students in all 3 courses:" ,(A&B&C))
print("students in at least 1 course:" ,(A|B|C))
print("students in exactly 1 course:" ,(A-(B|C))|(B-(A|C))|(C-(A|B)))
print("students in course 1 but not course 2:" ,(A-B) )    
print("students taking exactly 2 courses:" ,((A&B)-C) | ((B&C)-A) | ((A&C)-B))
```
