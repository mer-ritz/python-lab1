```python
#Find the smallest number in a list
a = [10, 5, 20, 3, 15]

smallest = min(a)

print("Smallest number:", smallest)
```

```python
#Find the average of numbers in a list
a = [10, 20, 30, 40, 50]

avg = sum(a) / len(a)

print("Average:", avg)
```

```python
#Count how many times a number occurs in a list
a = [1, 2, 3, 2, 4, 2, 5]

n = int(input("Enter number to count: "))

count = a.count(n)

print("Occurrences:", count)
```

```python
#Reverse a list
a = [10, 20, 30, 40, 50]

a.reverse()

print("Reversed list:", a)
```

```python
#Sort a list in ascending order
a = [40, 10, 30, 50, 20]

a.sort()

print("Sorted list:", a)
```

```python
#Find common elements of two tuples
a = (1, 2, 3, 4, 5)
b = (3, 4, 5, 6, 7)

common = tuple(x for x in a if x in b)

print("Common elements:", common)
```

```python
#Convert a list into a tuple
a = [10, 20, 30, 40, 50]

b = tuple(a)

print("Tuple:", b)
```

```python
#Convert a tuple into a list
a = (10, 20, 30, 40, 50)

b = list(a)

print("List:", b)
```

```python
#Find union, intersection and difference of two sets
A = {1, 2, 3, 4, 5}
B = {4, 5, 6, 7, 8}

print("Union:", A.union(B))
print("Intersection:", A.intersection(B))
print("Difference:", A.difference(B))
```

```python
#Remove duplicate values from a list using a set
a = [1, 2, 3, 2, 4, 3, 5, 1]

b = list(set(a))

print("List without duplicates:", b)
```
