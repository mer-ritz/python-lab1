```python
#Extract even numbers, calculate squares, and extract squares greater than 100#
a = [2, 4, 6, 8, 10, 12, 14]

even = [x for x in a if x % 2 == 0]
print("Even numbers:", even)

squares = [x**2 for x in even]
print("Squares:", squares)

result = [x for x in squares if x > 100]
print("Squares greater than 100:", result)
```

```python
#Extract strings with length greater than 3 and convert to uppercase
a = ["cat", "apple", "dog", "banana", "sun"]

result = [x.upper() for x in a if len(x) > 3]

print(result)
```

```python
#Extract numbers divisible by 2 or 3 and calculate their squares
a = [2, 3, 4, 5, 6, 7, 8, 9]

result = [x**2 for x in a if x % 2 == 0 or x % 3 == 0]

print(result)
```

```python
#Add 10 to every element and extract elements greater than 30
a = [10, 15, 20, 25, 30]

b = [x + 10 for x in a]
print("After adding 10:", b)

result = [x for x in b if x > 30]
print("Elements greater than 30:", result)
```

```python
#Extract negative numbers and convert them into absolute values
a = [-2, 5, -8, 10, -4, 6]

result = [abs(x) for x in a if x < 0]

print(result)
```

```python
#Extract words ending with 'n' and calculate their lengths
a = ["lion", "tiger", "elephant", "deer", "hen"]

result = [len(x) for x in a if x.endswith('n')]

print(result)
```

```python
#Extract perfect square numbers from a list
import math

a = [4, 9, 10, 16, 20, 25, 30, 36]

result = [x for x in a if math.isqrt(x)**2 == x]

print(result)
```
